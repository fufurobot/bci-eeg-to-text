```python
from pathlib import Path

import pandas as pd
sr = pd.Series(Path("data/eeg-files.txt").read_text().split("\n"))
df = pd.DataFrame(sr.apply(Path).apply(lambda x: x.parts).tolist())
df["suffix"] = sr.apply(Path).apply(lambda x: x.suffix)
df.to_csv("data/eeg-files.csv")
df
```
